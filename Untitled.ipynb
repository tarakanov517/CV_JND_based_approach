```python
ECC_MM_DATA = np.array([0.0, 0.25, 1.5, 4.0, 5.0, 9.0, 15.0, 17.0])
DENSITY_DATA = np.array([200000, 38000, 18000, 5000, 3900, 1000, 800, 400])

def load_model(config_path, use_gpu=False, use_jit=False):
    opt = load_config(config_path)
    model = eval(opt.Model.name)(**opt.Model)
    ckpt = os.path.join(opt.Test.Checkpoint.checkpoint_dir, 'latest.pth')
    if not os.path.exists(ckpt):
        ckpt = os.path.join(REPO_ROOT, 'latest.pth') 
        
    print(f"Loading weights from {ckpt}...")
    model.load_state_dict(torch.load(ckpt, map_location='cpu'), strict=True)
    
    if use_gpu: model = model.cuda()
    model.eval()
    
    if use_jit:
        jit_path = os.path.join(opt.Test.Checkpoint.checkpoint_dir, 'jit.pt')
        if not os.path.isfile(jit_path):
            try:
                model = Simplify(model)
                dummy = torch.rand(1, 3, *opt.Test.Dataset.transforms.static_resize.size)
                if use_gpu: dummy = dummy.cuda()
                model = torch.jit.trace(model, dummy, strict=False)
                torch.jit.save(model, jit_path)
            except NameError:
                print("Не найдено")
        else:
            del model
            model = torch.jit.load(jit_path)
    return model, opt

def pixels_to_ecc_mm(px_dist, img_w_px, screen_w_mm, view_dist_mm):
    mm_on_screen = (px_dist / img_w_px) * screen_w_mm
    ecc_deg = np.degrees(np.arctan(mm_on_screen / view_dist_mm))
    return ecc_deg * 0.28

def build_density_map(shape, center, config):
    """Creates density map centered at a specific point (cluster centroid)"""
    h, w = shape[:2]
    Y, X = np.ogrid[:h, :w]
    center_y, center_x = center[0], center[1] # (row, col)
    
    px_dist = np.sqrt((X - center_x)**2 + (Y - center_y)**2)
    
    ecc_mm = pixels_to_ecc_mm(px_dist, w, 
                               config['screen_width_mm'], config['viewing_distance_mm'])
    
    log_density = np.log10(DENSITY_DATA + 1)
    f_interp = interp1d(ECC_MM_DATA, log_density, kind='cubic', 
                        bounds_error=False, fill_value=(log_density[0], log_density[-1]))
    
    density_map = 10 ** f_interp(ecc_mm)
    
    d_min, d_max = density_map.min(), density_map.max()
    return (density_map - d_min) / (d_max - d_min + 1e-8)

def srgb_to_absolute_luminance(img_rgb, screen_lum):
    img = img_rgb.astype(np.float64) / 255.0
    lin = np.where(img <= 0.04045, img / 12.92, ((img + 0.055) / 1.055) ** 2.4)
    Y = 0.2126729 * lin[..., 0] + 0.7151522 * lin[..., 1] + 0.0721750 * lin[..., 2]
    return Y * screen_lum

def blur_at_centroid(image, centroid, area, sigma_base, sigma_scale):
    r, c = centroid
    sigma = sigma_base + sigma_scale * np.sqrt(area)
    Y, X = np.ogrid[:image.shape[0], :image.shape[1]]
    weight = np.exp(-((Y - r)**2 + (X - c)**2) / (2 * sigma**2))
    blurred = cv2.GaussianBlur(image.astype(np.float64), (0, 0), sigma)
    result = image.astype(np.float64) * (1 - weight[..., None]) + blurred * weight[..., None]
    return result.astype(np.uint8), sigma


def weighted_mean(lum_map, mask, density_map):
    w = density_map * mask.astype(np.float64)
    total = w.sum()
    return np.sum(lum_map * w) / total if total > 1e-8 else 0.0
```

```python
image_dir = os.path.join(REPO_ROOT, 'images_')
image_files = glob.glob(os.path.join(image_dir, '*.[jp][pn]g'))

try:
    model, opt = load_model(CONFIG['config_yaml'], CONFIG['use_gpu'], CONFIG['use_jit'])
except Exception as e:
    print(f"Ошибка загрузки модели: {e}")
    model, opt = None, None

results_summary = []

for img_path in image_files:
    
    if model:
        try:
            loader = ImageLoader(img_path, opt.Test.Dataset.transforms)
            sample = next(iter(loader))
            
            if CONFIG['use_gpu']:
                sample = to_cuda(sample)
                
            with torch.no_grad():
                out = model(sample)
                
            pred = to_numpy(out['pred'], sample['shape'])
            if isinstance(pred, list): pred = pred[0]
            if len(pred.shape) == 3: pred = pred.squeeze(0)
            
            original_img = np.array(sample['original'])
            saliency_map = pred
            
        except Exception as e:
            print(f"Ошибка инференса: {e}")
            continue
    else:
        img_dummy = cv2.imread(img_path)
        original_img = cv2.cvtColor(img_dummy, cv2.COLOR_BGR2RGB)
        saliency_map = np.random.rand(original_img.shape[0], original_img.shape[1])

    # DBSCAN
    saliency_uint8 = (saliency_map * 255).astype(np.uint8)
    _, binary_mask = cv2.threshold(saliency_uint8, 0, 255, cv2.THRESH_BINARY | cv2.THRESH_OTSU)

    points = np.column_stack(np.where(binary_mask > 0)) 
    step = 5
    sampled_points = points[::step]

    if len(sampled_points) < CONFIG['min_roi_area']:
        final_la = 0.0
        cluster_results = []
        unique_labels = set()
    else:
        db = DBSCAN(eps=15, min_samples=CONFIG['min_roi_area']).fit(sampled_points)
        labels = db.labels_
        unique_labels = set(labels)
        unique_labels.discard(-1)

        print(f"Найдено {len(unique_labels)} кластеров (AOIs)")

        cluster_results = []
        la_before_list = []; la_after_raw_list = []; la_after_weighted_list = []
        
        for label in unique_labels:
            cluster_points = sampled_points[labels == label]
            if len(cluster_points) == 0: continue
                
            centroid = np.mean(cluster_points, axis=0)
            area = len(cluster_points) * (step ** 2)
            
            mask = np.zeros(binary_mask.shape, dtype=np.uint8)
            for pt in cluster_points:
                cv2.circle(mask, (int(pt[1]), int(pt[0])), 5, 255, -1)
            
            blurred_img, sigma = blur_at_centroid(
                original_img, centroid, area, 
                CONFIG['blur_sigma_base'], CONFIG['blur_sigma_scale']
            )
            
            lum_map = srgb_to_absolute_luminance(blurred_img, CONFIG['screen_luminance_cd_m2'])
            density_map = build_density_map(original_img.shape, centroid, CONFIG)
            la_cluster = weighted_mean(lum_map, mask > 0, density_map)
            
            lum_orig = srgb_to_absolute_luminance(original_img, CONFIG['screen_luminance_cd_m2'])
            la_before_list.append(la_cluster)
            
            blurred_img, _ = blur_at_centroid(original_img, centroid, area, CONFIG['blur_sigma_base'], CONFIG['blur_sigma_scale'])
            
            lum_blur_raw = srgb_to_absolute_luminance(blurred_img, CONFIG['screen_luminance_cd_m2'])
            la_after_raw = np.mean(lum_blur_raw[mask > 0])
            la_after_raw_list.append(la_after_raw)
            
            la_after_weighted = weighted_mean(lum_blur_raw, mask > 0, density_map)
            la_after_weighted_list.append(la_after_weighted)
            
            cluster_results.append({
                'label': label, 'centroid': centroid, 
                'blurred_img': blurred_img, 
                'la_before': la_cluster, 
                'la_after_raw': la_after_raw,
                'la_after_weighted': la_after_weighted
            })

        final_la = np.mean(la_after_weighted_list) if cluster_results else 0.0
        
        print(f"Средняя яркость изображения (Y): {np.mean(lum_orig):.2f} кд/м²")
        # print(f"Яркость кластеров ДО размытия (взвеш.):       {[f'{x:.2f}' for x in la_before_list]}")
        print(f"Яркость кластеров после размытия:     {[f'{x:.2f}' for x in la_after_raw_list]}")
        print(f"Яркость кластеров после размытия (взвеш.):    {[f'{x:.2f}' for x in la_after_weighted_list]}")
        print(f"Итоговая La: {final_la:.2f} кд/м²")
        

    print(f"Яркость адаптации (La): {final_la:.4f} кд/м²")
    
    results_summary.append({'file': os.path.basename(img_path), 'la': final_la})

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].imshow(original_img)
    axes[0].imshow(saliency_map, cmap='jet', alpha=0.5)
    axes[0].set_title("InSPyReNet Saliency")
    axes[0].axis('off')

    axes[1].imshow(binary_mask, cmap='gray')
    if len(sampled_points) >= CONFIG['min_roi_area']:
        for lbl in unique_labels:
            m = labels == lbl
            axes[1].scatter(sampled_points[m, 1], sampled_points[m, 0], s=2, c='r')
    axes[1].set_title(f"DBSCAN кластеры ({len(unique_labels)})")
    axes[1].axis('off')

    if cluster_results:
        res = cluster_results[0]
        axes[2].imshow(res['blurred_img'])
        r, c = res['centroid']
        axes[2].plot(c, r, 'y+', markersize=15, markeredgewidth=2)
        axes[2].set_title("Центр масс кластера")
        
    axes[2].axis('off')

    plt.suptitle(f"{os.path.basename(img_path)} | La: {final_la:.2f}", fontsize=14)
    plt.tight_layout()
    plt.show()

os.chdir(original_cwd)
```

## cv2.saliency (SpectralResidual) + DBSCAN

```python
image_dir = Path("images_")
image_files = list(image_dir.glob("*.[jp][pn]g")) + list(image_dir.glob("*.[jp][pn]G"))
```

```python
for img_path in image_files:
    print(f"Файл: {os.path.basename(img_path)}")
    
    img_bgr = cv2.imread(str(img_path))
    if img_bgr is None: continue
    original_img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    
    img_xyz = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2XYZ)
    mean_img_y = np.mean(img_xyz[..., 1]) / 255.0 * CONFIG['screen_luminance_cd_m2']

    saliency_obj = cv2.saliency.StaticSaliencySpectralResidual_create()
    _, saliency_map_float = saliency_obj.computeSaliency(img_bgr)
    saliency_map_uint8 = (saliency_map_float * 255).astype(np.uint8)
    _, binary_mask = cv2.threshold(saliency_map_uint8, 0, 255, cv2.THRESH_BINARY | cv2.THRESH_OTSU)

    points = np.column_stack(np.where(binary_mask > 0)) 
    step = 5
    sampled_points = points[::step]

    if len(sampled_points) < CONFIG['min_roi_area']:
        final_la = 0.0; cluster_results = []; unique_labels = set(); labels = np.array([])
    else:
        db = DBSCAN(eps=15, min_samples=CONFIG['min_roi_area']).fit(sampled_points)
        labels = db.labels_
        unique_labels = set(labels); unique_labels.discard(-1)

        cluster_results = []
        la_before_list = []; la_after_raw_list = []; la_after_weighted_list = []

        for label in unique_labels:
            cluster_points = sampled_points[labels == label]
            if len(cluster_points) == 0: continue
            
            centroid = np.mean(cluster_points, axis=0)
            area = len(cluster_points) * (step ** 2)
            
            mask = np.zeros(binary_mask.shape, dtype=np.uint8)
            for pt in cluster_points: cv2.circle(mask, (int(pt[1]), int(pt[0])), 5, 255, -1)
            
            dens_map = build_density_map(original_img.shape, centroid, CONFIG)
            
            lum_orig = srgb_to_absolute_luminance(original_img, CONFIG['screen_luminance_cd_m2'])
            la_before = weighted_mean(lum_orig, mask > 0, dens_map)
            la_before_list.append(la_before)
            
            blurred_img, _ = blur_at_centroid(original_img, centroid, area, CONFIG['blur_sigma_base'], CONFIG['blur_sigma_scale'])
            
            lum_blur_raw = srgb_to_absolute_luminance(blurred_img, CONFIG['screen_luminance_cd_m2'])
            la_after_raw = np.mean(lum_blur_raw[mask > 0])
            la_after_raw_list.append(la_after_raw)
            
            la_after_weighted = weighted_mean(lum_blur_raw, mask > 0, dens_map)
            la_after_weighted_list.append(la_after_weighted)
            
            cluster_results.append({
                'label': label, 'centroid': centroid, 
                'blurred_img': blurred_img, 
                'la_before': la_before, 
                'la_after_raw': la_after_raw,
                'la_after_weighted': la_after_weighted
            })

        final_la = np.mean(la_after_weighted_list) if cluster_results else 0.0
        
        print(f"Средняя яркость изображения (Y): {mean_img_y:.2f} кд/м²")
        print(f"Яркость кластеров после размытия:     {[f'{x:.2f}' for x in la_after_raw_list]}")
        print(f"Яркость кластеров ПОСЛЕ размытия (взвеш.):    {[f'{x:.2f}' for x in la_after_weighted_list]}")
        print(f"Итоговая La: {final_la:.2f} кд/м²")

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    
    axes[0].imshow(original_img)
    axes[0].imshow(saliency_map_float, cmap='jet', alpha=0.5)
    axes[0].set_title("Spectral Residual Saliency")
    axes[0].axis('off')

    axes[1].imshow(binary_mask, cmap='gray')
    if cluster_results:
        for res in cluster_results:
            m = labels == res['label']
            axes[1].scatter(sampled_points[m, 1], sampled_points[m, 0], s=2, c='r')
            r, c = res['centroid']
            txt = f"{res['la_before']:.0f}/{res['la_after_weighted']:.0f}"
            axes[1].text(c, r, txt, color='yellow', fontsize=9, fontweight='bold', 
                         ha='center', va='center', bbox=dict(facecolor='black', alpha=0.7, pad=1, edgecolor='none'))
    
    axes[1].set_title(f"DBSCAN ({len(unique_labels)}) | Before/After")
    axes[1].axis('off')

    plt.suptitle(f"Spectral Residual | La: {final_la:.2f} | Img Y: {mean_img_y:.2f}", fontsize=14)
    plt.tight_layout()
    plt.show()
```
